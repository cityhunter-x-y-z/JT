```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from datetime import datetime, timedelta
from google.colab import auth
import gspread
from google.auth import default

# Authenticate
auth.authenticate_user()
creds, _ = default()
gc = gspread.authorize(creds)

print("✅ Packages loaded and authenticated!")
```

```python
# Connect to your Google Sheet
SHEET_NAME = 'Job Hunt Tracker'  # Replace with your sheet name

spreadsheet = gc.open(SHEET_NAME)

# Load all sheets
job_tracker = pd.DataFrame(spreadsheet.worksheet('Job Tracker').get_all_records())
weekly_metrics = pd.DataFrame(spreadsheet.worksheet('Weekly Metrics').get_all_records())
application_sources = pd.DataFrame(spreadsheet.worksheet('Application Sources').get_all_records())

print(f"✅ Loaded data from '{SHEET_NAME}'")
print(f"\nJobs tracked: {len(job_tracker)}")
print(f"Weeks tracked: {len(weekly_metrics)}")
```

## 📊 Current Status Overview

```python
# Current week status
print("=" * 60)
print("CURRENT WEEK STATUS")
print("=" * 60)

# Jobs by status
status_counts = job_tracker['Status'].value_counts()
print("\n📌 Jobs by Status:")
for status, count in status_counts.items():
    print(f"  {status}: {count}")

# Priority breakdown
priority_counts = job_tracker['Priority'].value_counts()
print("\n⭐ Jobs by Priority:")
for priority, count in priority_counts.items():
    print(f"  Priority {priority}: {count}")

# Response rate
if len(weekly_metrics) > 0:
    latest_week = weekly_metrics.iloc[-1]
    print("\n📈 Latest Week Metrics:")
    print(f"  Applications Sent: {latest_week['Applications Sent']}")
    print(f"  Direct Emails: {latest_week['Direct Emails Sent']}")
    print(f"  Responses: {latest_week['Responses Received']}")
    print(f"  Response Rate: {latest_week['Response Rate']}")
    print(f"  Phone Screens: {latest_week['Phone Screens']}")
    print(f"  Interviews: {latest_week['Interviews']}")
```

## 🎯 Priority Jobs This Week

```python
# Show Priority A jobs that need action
priority_a = job_tracker[
    (job_tracker['Priority'] == 'A') & 
    (job_tracker['Status'].isin(['Researching', 'Not contacted']))
].copy()

print("🎯 PRIORITY A JOBS - NEED ACTION")
print("=" * 60)

if len(priority_a) > 0:
    for idx, job in priority_a.head(10).iterrows():
        print(f"\n{job['ID']}. {job['Company']} - {job['Job Title']}")
        print(f"   Status: {job['Status']}")
        print(f"   Hiring Manager: {job.get('Hiring Manager', 'TO FIND')}")
        print(f"   Email: {job.get('Email', 'TO FIND')}")
        print(f"   Notes: {job.get('Notes', 'Need to research')}")
else:
    print("\n✅ No Priority A jobs pending!")
    print("   → Add more priority jobs to tracker")
    print("   → Or focus on Priority B jobs")
```

## 📧 Follow-ups Needed

```python
# Check which applications need follow-up
# Assuming you have a 'Follow-up Date' column in your tracker

today = datetime.now().date()

# Convert Follow-up Date column if it exists
if 'Follow-up Date' in job_tracker.columns:
    job_tracker['Follow-up Date'] = pd.to_datetime(job_tracker['Follow-up Date'], errors='coerce')
    
    follow_ups_needed = job_tracker[
        (job_tracker['Follow-up Date'].notna()) &
        (job_tracker['Follow-up Date'].dt.date <= today) &
        (job_tracker['Status'].isin(['Applied', 'Followed Up']))
    ].copy()
    
    print("📧 FOLLOW-UPS NEEDED THIS WEEK")
    print("=" * 60)
    
    if len(follow_ups_needed) > 0:
        for idx, job in follow_ups_needed.iterrows():
            print(f"\n{job['Company']} - {job['Job Title']}")
            print(f"   Applied: {job.get('Applied Date', 'Unknown')}")
            print(f"   Follow-up due: {job['Follow-up Date'].strftime('%Y-%m-%d')}")
            print(f"   Email: {job.get('Email', 'Unknown')}")
    else:
        print("\n✅ No follow-ups due this week!")
else:
    print("ℹ️ Add 'Follow-up Date' column to track follow-ups")
```

## 📊 Performance Analysis

```python
# Analyze your performance over weeks
if len(weekly_metrics) > 0:
    print("📊 WEEKLY PERFORMANCE TREND")
    print("=" * 60)
    
    # Plot applications vs responses
    plt.figure(figsize=(12, 5))
    
    plt.subplot(1, 2, 1)
    plt.plot(weekly_metrics['Week Starting'], weekly_metrics['Applications Sent'], marker='o', label='Applications')
    plt.plot(weekly_metrics['Week Starting'], weekly_metrics['Responses Received'], marker='o', label='Responses')
    plt.xlabel('Week')
    plt.ylabel('Count')
    plt.title('Applications vs Responses')
    plt.legend()
    plt.xticks(rotation=45)
    
    plt.subplot(1, 2, 2)
    plt.plot(weekly_metrics['Week Starting'], weekly_metrics['Response Rate'].str.rstrip('%').astype(float), marker='o', color='green')
    plt.xlabel('Week')
    plt.ylabel('Response Rate (%)')
    plt.title('Response Rate Trend')
    plt.xticks(rotation=45)
    plt.axhline(y=10, color='r', linestyle='--', label='10% Target')
    plt.legend()
    
    plt.tight_layout()
    plt.show()
    
    # Summary stats
    total_apps = weekly_metrics['Applications Sent'].sum()
    total_responses = weekly_metrics['Responses Received'].sum()
    overall_response_rate = (total_responses / total_apps * 100) if total_apps > 0 else 0
    
    print(f"\n📈 OVERALL STATS:")
    print(f"   Total Applications: {total_apps}")
    print(f"   Total Responses: {total_responses}")
    print(f"   Overall Response Rate: {overall_response_rate:.1f}%")
    print(f"   Total Interviews: {weekly_metrics['Interviews'].sum()}")
    print(f"   Total Offers: {weekly_metrics['Offers'].sum()}")
    
    # Recommendations
    print("\n💡 RECOMMENDATIONS:")
    if overall_response_rate < 5:
        print("   ⚠️ Response rate is low (<5%)")
        print("      → Review your resume and portfolio")
        print("      → Are you personalizing applications?")
        print("      → Focus on quality over quantity")
    elif overall_response_rate < 10:
        print("   📈 Response rate is okay (5-10%)")
        print("      → Keep going, you're on track")
        print("      → Try more direct outreach to hiring managers")
    else:
        print("   ✅ Great response rate (>10%)!")
        print("      → Your approach is working")
        print("      → Consider increasing volume slightly")
else:
    print("ℹ️ No metrics data yet - start tracking this week!")
```

## 🎯 Application Source Analysis

```python
# Analyze which sources work best
if len(application_sources) > 0 and application_sources['Applications'].sum() > 0:
    print("🎯 BEST APPLICATION SOURCES")
    print("=" * 60)
    
    sources_with_data = application_sources[application_sources['Applications'] > 0].copy()
    sources_with_data['Response Rate'] = sources_with_data['Response Rate'].str.rstrip('%').astype(float)
    sources_with_data = sources_with_data.sort_values('Response Rate', ascending=False)
    
    print("\nSource Performance:")
    for idx, source in sources_with_data.iterrows():
        print(f"\n{source['Source']}:")
        print(f"   Applications: {source['Applications']}")
        print(f"   Responses: {source['Responses']}")
        print(f"   Response Rate: {source['Response Rate']:.1f}%")
        print(f"   Interviews: {source['Interviews']}")
    
    print("\n💡 FOCUS YOUR ENERGY:")
    best_source = sources_with_data.iloc[0]
    print(f"   → {best_source['Source']} has your best response rate ({best_source['Response Rate']:.1f}%)")
    print(f"   → Prioritize this source for applications")
else:
    print("ℹ️ Start tracking application sources to see what works best!")
```

## 🎓 Tips for Success

### Time Management
- **Morning (1 hour):** Research 3 companies, add to tracker
- **Afternoon (1 hour):** Customize resume, send 2 applications
- **Evening (30 mins):** LinkedIn activity, follow-ups

### Quality Indicators
- Spending 15-20 mins researching each company? ✅
- Personalizing every email? ✅
- Mentioning specific products/features in applications? ✅
- Following up 5-7 days after applying? ✅

### Red Flags
- Sending 50+ applications per week? ⚠️ Too generic
- Response rate < 3%? ⚠️ Not personalizing enough
- Not tracking anything? ⚠️ Can't improve what you don't measure

### Motivation
- **Week 1-2:** Setup and learning (don't expect results yet)
- **Week 3-4:** Responses start coming in (keep going!)
- **Week 5-8:** Phone screens and interviews (refine your approach)
- **Week 8-12:** Offers (most people get offers in this range)

Job search is a numbers game + strategy. You're building a system that works. Trust the process! 💪
