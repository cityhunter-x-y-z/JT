```python
# YOUR INFORMATION - Fill this out once

YOUR_INFO = {
    'name': 'Your Full Name',
    'email': 'your.email@gmail.com',
    'phone': '+91-XXXXX-XXXXX',
    'linkedin': 'https://linkedin.com/in/yourprofile',
    'portfolio': 'https://yourportfolio.com',
    'location': 'Bangalore, India',
    'years_experience': '3',
    'specialty': 'Mobile-first Product Design',  # e.g., "Fintech UX", "B2B SaaS", etc.
    'current_company': 'Your Current Company',
    'top_skills': ['User Research', 'Prototyping', 'Design Systems']  # Top 3-5 skills
}

# Your top achievements (with metrics)
ACHIEVEMENTS = [
    'Redesigned checkout flow, increasing conversion by 23%',
    'Led design system adoption across 4 product teams',
    'Conducted 50+ user interviews to validate product direction'
]

print("✅ Your information loaded!")
```

```python
# Connect to Google Sheets
from google.colab import auth
import gspread
from google.auth import default
import pandas as pd

# Authenticate
auth.authenticate_user()
creds, _ = default()
gc = gspread.authorize(creds)

# Open your spreadsheet (replace with your sheet name)
SHEET_NAME = 'Job Hunt Tracker'  # Replace with your Google Sheet name

try:
    spreadsheet = gc.open(SHEET_NAME)
    print(f"✅ Connected to '{SHEET_NAME}'")
    
    # Load job tracker
    job_sheet = spreadsheet.worksheet('Job Tracker')
    jobs_df = pd.DataFrame(job_sheet.get_all_records())
    
    print(f"\n📊 Loaded {len(jobs_df)} jobs from tracker")
    print("\nRecent jobs:")
    print(jobs_df[['Company', 'Job Title', 'Priority', 'Status']].head(10))
    
except Exception as e:
    print(f"❌ Error: {e}")
    print("\nMake sure:")
    print("1. You've created a Google Sheet named 'Job Hunt Tracker'")
    print("2. The sheet has a tab named 'Job Tracker'")
    print("3. You've shared the sheet with your Google account")
```

```python
# Email Template Functions

def generate_cold_outreach(job_data, research_notes):
    """
    Generate personalized cold outreach email.
    
    Args:
        job_data: Row from jobs_df with company info
        research_notes: Dict with 'specific_aspect', 'specific_reason', 'relevant_project'
    """
    
    subject = f"Product Designer interested in {job_data['Company']}'s {research_notes.get('product', 'products')}"
    
    body = f"""Hi {job_data['Hiring Manager']},

I came across {job_data['Company']}'s {job_data['Job Title']} role and was immediately drawn to {research_notes['specific_aspect']}.

I'm a Product Designer based in {YOUR_INFO['location']} with {YOUR_INFO['years_experience']} years of experience in {YOUR_INFO['specialty']}. What excites me about {job_data['Company']} is {research_notes['specific_reason']}.

In my current role at {YOUR_INFO['current_company']}, I:
• {ACHIEVEMENTS[0]}
• {ACHIEVEMENTS[1]}
• {ACHIEVEMENTS[2]}

I've put together a case study that might be relevant: {research_notes['relevant_project']} - it shows how I {research_notes.get('project_skill', 'approach similar challenges')}.

I'd love to chat about how my experience in {YOUR_INFO['specialty']} could contribute to {research_notes.get('their_goal', 'your team')}. Are you the right person to discuss this, or could you point me to who I should reach out to?

My portfolio: {YOUR_INFO['portfolio']}

Thanks for your time,
{YOUR_INFO['name']}
{YOUR_INFO['phone']}
{YOUR_INFO['linkedin']}"""
    
    return subject, body


def generate_follow_up(job_data, days_since_last_email, new_info=None):
    """
    Generate follow-up email.
    """
    
    subject = f"Re: Product Designer interested in {job_data['Company']}'s {job_data.get('Product', 'products')}"
    
    time_reference = "last week" if days_since_last_email > 5 else "a few days ago"
    
    body = f"""Hi {job_data['Hiring Manager']},

Following up on my email from {time_reference}. I understand you're likely busy, so I wanted to resurface this briefly.

I'm genuinely excited about {job_data['Company']}'s work, especially {job_data.get('Notes', 'your recent updates')}.
"""
    
    if new_info:
        body += f"\n{new_info}\n"
    
    body += f"""
Would you have 15 minutes for a quick chat about the {job_data['Job Title']} role? Happy to work around your schedule.

Thanks,
{YOUR_INFO['name']}"""
    
    return subject, body


print("✅ Email template functions loaded!")
```

```python
# GENERATE EMAIL FOR A SPECIFIC JOB

# Step 1: Select a job by ID or company name
JOB_ID = 1  # Change this to the job ID from your tracker

# Or select by company name:
# COMPANY_NAME = 'Example Corp'
# selected_job = jobs_df[jobs_df['Company'] == COMPANY_NAME].iloc[0]

selected_job = jobs_df[jobs_df['ID'] == JOB_ID].iloc[0]

print("Selected job:")
print(f"Company: {selected_job['Company']}")
print(f"Role: {selected_job['Job Title']}")
print(f"Hiring Manager: {selected_job.get('Hiring Manager', 'Unknown')}")
print(f"Email: {selected_job.get('Email', 'Unknown')}")
print(f"\nNotes from tracker: {selected_job.get('Notes', 'None')}")
```

```python
# Step 2: Add your research notes (CUSTOMIZE THIS!)

research_notes = {
    'product': 'mobile app',  # Specific product they build
    'specific_aspect': 'your focus on building delightful experiences for Tier 2/3 cities',
    'specific_reason': 'the challenge of designing for diverse user contexts and connectivity conditions',
    'relevant_project': 'https://yourportfolio.com/project/mobile-first-design',  # Link to specific case study
    'project_skill': 'design for low-bandwidth environments',
    'their_goal': 'expanding to new markets'
}

print("✅ Research notes added!")
print("\n⚠️ Make sure you've customized the research_notes dict above!")
```

```python
# Step 3: Generate the email

subject, body = generate_cold_outreach(selected_job, research_notes)

print("="*60)
print("GENERATED EMAIL")
print("="*60)
print(f"\nTO: {selected_job.get('Email', 'EMAIL_NOT_FOUND')}")
print(f"SUBJECT: {subject}")
print(f"\n{body}")
print("\n" + "="*60)
print("\n📋 Copy the email above and paste into Gmail")
print("⚠️ REVIEW AND EDIT before sending!")
print("✅ Then mark as 'Applied' in your tracker")
```

```python
# Optional: Generate follow-up email

days_since_last_email = 7  # How many days since you sent the first email
new_info = "Since my last email, I completed a case study on mobile-first design: https://yourportfolio.com/new-project"  # Optional

subject_followup, body_followup = generate_follow_up(selected_job, days_since_last_email, new_info)

print("="*60)
print("FOLLOW-UP EMAIL")
print("="*60)
print(f"\nTO: {selected_job.get('Email', 'EMAIL_NOT_FOUND')}")
print(f"SUBJECT: {subject_followup}")
print(f"\n{body_followup}")
print("\n" + "="*60)
```

```python
# BATCH VIEW: See all Priority A jobs that need outreach

priority_jobs = jobs_df[
    (jobs_df['Priority'] == 'A') & 
    (jobs_df['Status'].isin(['Researching', 'Not contacted']))
].copy()

print(f"📊 {len(priority_jobs)} Priority A jobs need outreach:\n")
for idx, job in priority_jobs.iterrows():
    print(f"ID: {job['ID']} | {job['Company']} - {job['Job Title']}")
    print(f"   Hiring Manager: {job.get('Hiring Manager', 'Unknown')}")
    print(f"   Email: {job.get('Email', 'TO FIND')}")
    print()
```

## 📝 Usage Instructions

### For Each Job Application:

1. **Research the company** (15-20 mins)
   - Read their product pages
   - Check recent news/blog posts
   - Look at their design team on LinkedIn
   - Find specific features/projects to mention

2. **Find hiring manager**
   - LinkedIn: Search "[Company] design manager"
   - Company website: Team page
   - LinkedIn: Company page → People tab → filter by Design

3. **Get email address**
   - Hunter.io (25 free searches/month)
   - Company website contact page
   - Common patterns: firstname@company, first.last@company

4. **Update research_notes cell**
   - Be specific! Generic = ignored
   - Mention actual products/features
   - Link to relevant portfolio project

5. **Generate, review, customize, send**
   - Copy generated email
   - Paste to Gmail
   - Read through and add personal touches
   - Send during business hours (10 AM - 4 PM IST)

6. **Update your tracker**
   - Mark status as "Applied"
   - Set follow-up date (7 days later)
   - Track in metrics sheet

### Daily Routine:
- Morning: Research 3-5 new jobs, add to tracker
- Afternoon: Generate and send 2-3 personalized emails
- Evening: Follow up on pending applications (if 5-7 days old)

### Weekly Review:
- Check response rate
- Adjust template if response rate < 10%
- Celebrate responses!
- Plan next week's targets
